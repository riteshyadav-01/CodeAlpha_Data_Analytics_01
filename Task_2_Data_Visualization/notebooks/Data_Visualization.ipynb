```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
df = pd.read_csv("../dataset/cleaned_womens_clothing_reviews.csv")
```

```python
plt.figure(figsize=(8, 5))

sns.countplot(
    data=df,
    x='Division Name',
    order=df['Division Name'].value_counts().index
)

plt.title('Division-wise Customer Reviews')
plt.xlabel('Division')
plt.ylabel('Number of Reviews')

plt.tight_layout()
plt.savefig('../charts/division_reviews.png')
plt.show()
```

```python
plt.figure(figsize=(9, 5))

sns.histplot(data=df, x='Age', bins=20, kde=True)

plt.title('Customer Age Distribution')
plt.xlabel('Customer Age')
plt.ylabel('Number of Customers')

plt.tight_layout()
plt.savefig('../charts/age_distribution.png')
plt.show()
```

```python
plt.figure(figsize=(7, 5))

sns.countplot(
    data=df,
    x='Recommended IND'
)

plt.title('Customer Recommendation Distribution')
plt.xlabel('Recommended (0 = No, 1 = Yes)')
plt.ylabel('Number of Customers')

plt.tight_layout()
plt.savefig('../charts/recommendation_distribution.png')
plt.show()
```

```python
plt.figure(figsize=(9, 5))

sns.barplot(
    data=df,
    x='Department Name',
    y='Rating',
    estimator='mean',
    errorbar=None
)

plt.title('Average Rating by Department')
plt.xlabel('Department')
plt.ylabel('Average Rating')
plt.xticks(rotation=30)

plt.tight_layout()
plt.savefig('../charts/department_average_rating.png')
plt.show()
```

```python
top_products = df['Clothing ID'].value_counts().head(10)

plt.figure(figsize=(10, 5))

sns.barplot(
    x=top_products.index.astype(str),
    y=top_products.values
)

plt.title('Top 10 Most Reviewed Products')
plt.xlabel('Clothing ID')
plt.ylabel('Number of Reviews')
plt.xticks(rotation=45)

plt.tight_layout()
plt.savefig('../charts/top_10_products.png')
plt.show()
```

```python
plt.figure(figsize=(10, 5))

sns.boxplot(
    data=df,
    x='Department Name',
    y='Rating'
)

plt.title('Rating Distribution by Department')
plt.xlabel('Department')
plt.ylabel('Customer Rating')
plt.xticks(rotation=30)

plt.tight_layout()
plt.savefig('../charts/rating_by_department.png')
plt.show()
```

```python
plt.figure(figsize=(9, 6))

numeric_df = df.select_dtypes(include='number')
correlation = numeric_df.corr()

sns.heatmap(
    correlation,
    annot=True,
    cmap='coolwarm',
    fmt='.2f'
)

plt.title('Correlation Heatmap of Customer Reviews')

plt.tight_layout()
plt.savefig('../charts/correlation_heatmap.png')
plt.show()
```

```python
print("DATA VISUALIZATION SUMMARY")
print("-" * 40)

print("\n1. Total Reviews:", len(df))

print("\n2. Most Reviewed Department:")
print(df['Department Name'].value_counts().idxmax())

print("\n3. Most Common Rating:")
print(df['Rating'].mode()[0])

print("\n4. Most Recommended Customers:")
print(df['Recommended IND'].value_counts()[1])

print("\n5. Average Customer Age:")
print(round(df['Age'].mean(), 2))

print("\n6. Average Rating:")
print(round(df['Rating'].mean(), 2))

print("\nTask 2 Visualizations Completed Successfully!")
```
